## Series

Start by creating four new variables named aa, ab, ac, and ad to hold some plain
Python objects. 
- Assign aa the list 1,2,3,4.
- Assign ab the tuple 5,6,7,8.
- Assign ac the set 9,10,11,12.
- Assign ad a dictionary with the following key-value pairs:\
'first': 13, 'second': 14, 'third': 15, 'fourth': 16

```python
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {
    'first': 13,
    'second': 14,
    'third': 15,
    'fourth': 16
}
```

Create four new variables named ser1, ser2, ser3, and ser4. Assign to each a
Series created from one of the previous four variables.
- ser1 should use aa
- ser2 should use ab
- ser3 should use ac
- ser4 should use ad\
Print each of new Series to see the result.

```python
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(list(ac)) # Gives an TypeError without casting as list because the set is unordered 
ser4 = pd.Series(ad)

print(ser1)
print(ser2)
print(ser3)
print(ser4)
```

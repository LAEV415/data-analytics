# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

Create a DataFrame named df_sales using the following data:\
data = {'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],\
'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East',
'West', 'North'],\
'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],\
'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]}

```python
data = {'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]}
df_sales = pd.DataFrame(data)
```

Add a new column Revenue to df_sales by multiplying Units_Sold and Price.

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

Use the sum function to calculate the total Units_Sold and total Revenue across all
products and regions.

```python
print('Units Sold:', np.sum(df_sales['Units_Sold']))
print('Total Revenue:', np.sum(df_sales['Revenue']))
```

Use the mean function to calculate the average Units_Sold and average Revenue.

```python
print(f'Average Units Sold: {round(df_sales['Units_Sold'].mean(), 2)}')
print(f'Average Revenue: {round(df_sales['Revenue'].mean(), 2)}')
```

## Grouping Data

Use the groupby function to group df_sales by Product and calculate the total
Units_Sold and total Revenue for each product.

```python
prod_sales = df_sales.groupby(['Product']).aggregate({
    'Units_Sold': 'sum',
    'Revenue': 'sum'
})
prod_sales
```

Use the groupby function to group df_sales by Region and calculate the average
Units_Sold and average Revenue for each region.

```python
reg_sales = df_sales.groupby(['Region']).aggregate({
    'Units_Sold': 'mean',
    'Revenue': 'mean'
})
reg_sales
```

Use the groupby function to group df_sales by both Product and Region and
calculate the sum of Units_Sold and Revenue for each combination.

```python
prod_reg_sales = df_sales.groupby(['Product','Region']).aggregate({
    'Units_Sold': 'sum',
    'Revenue': 'sum'    
})
prod_reg_sales
```

## Sorting and Ranking Data

Sort df_sales by Units_Sold in ascending order.

```python
asc_sales = df_sales.sort_values('Units_Sold')
asc_sales
```

Sort df_sales by Revenue in descending order.

```python
desc_sales = df_sales.sort_values('Revenue', ascending=False)
desc_sales
```

Add a new column Revenue_Rank to df_sales, which ranks the Revenue values in
descending order (i.e., highest revenue has rank 1).

```python
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False)
df_sales = df_sales.sort_values('Revenue', ascending=False)
df_sales
```

## Merging DataFrames

Create a new DataFrame named df_products with the following data:\
product_data = {'Product': ['A', 'B', 'C'],\
'Category': ['Electronics', 'Furniture', 'Appliances']}\
df_products = pd.DataFrame(product_data)

```python
product_data = {'Product': ['A', 'B', 'C'],
'Category': ['Electronics', 'Furniture', 'Appliances']}
df_products = pd.DataFrame(product_data)
df_products
```

Merge df_sales with df_products on the Product column, resulting in a new
DataFrame named df_merged.

```python
df_merged = pd.merge(df_sales, df_products, on=['Product'])
```

Call df_merged to review the merged DataFrame.

```python
df_merged
```

```python
df_merged = df_merged.groupby(['Category']).aggregate({'Revenue': 'sum'})
df_merged
```

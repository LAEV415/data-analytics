## Generating Random Numbers

Create a variable named r_1 that contains a random set of integers between 1 and
20 with a size of 2 elements. Print r_1.

```python
r_1 = np.random.randint(1, 20, size=(2))
print(r_1)
```

Create a variable named r_2 that contains a random set of integers between 1 and
20 with a size of 1 row, 2 columns. Print r_2.

```python
r_2 = np.random.randint(1, 20, size=(1, 2))
print(r_2)
```

Create a variable named r_3 that contains a random set of floats between 0 and 1
with a size of 2 rows, 1 columns. Print r_3.

```python
r_3 = np.random.rand(2, 1)
print(r_3)
```

Create a variable named r_4 that contains a random set of integers between 20
and 100 with a size of 4 rows, 5 columns. Print r_4.

```python
r_4 = np.random.randint(20, 100, size=(4, 5))
print(r_4)
```

Create a variable named r_5 that contains a random set of floats between 0 and 1
with a size of 4 rows, 5 columns. Print r_5.

```python
r_5 = np.random.rand(4, 5)
print(r_5)
```

## Generating Random Choices from an Array

Create a variable named students and assign it a 1-D array containing the names
of 12 of your classmates.

```python
students = np.array(['Edmund', 'Brandon', 'Kevin', 'Dahlia', 'Natalia', 'Destiny', 'Jeremiah', 'Hritik', 'Stefhanie', 'Luis', 'Bess', 'James Bond'])
```

Create a variable named call_on and use it to choose one random selection from
students. Print call_on.

```python
call_on = np.random.choice(students)
print(call_on)
```

Create a variable named breakout_groups and use it to choose a random
sampling from students that is 3 rows, 4 columns in size. Print breakout_groups and examine the results – would this be an effective way to
assign breakout groups? Why or why not? Add a comment to the cell with your
response.

```python
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups) 
# This is not an effective way to assign breakout groups because for each list, 
# names are repeatedly used and we can not have one person in separate meetings.
```

Create a variable named call_on_weighted and use it to choose make five
random selections from students, but this time assign different probability values
that will determine how likely each student is to be selected. At least one student
should have a zero likelihood of being called on, and at least one student should
have a 30% likelihood of being called on. Distribute other probability values as you
choose, and in a comment, explain the p values you assigned to each student. Print
call_on_weighted.

```python
call_on_weighted = np.random.choice(students, 
                                    p=[.05, .05, .05, .05, .05, .05, .1, .1, .1, 0, .1, .3]
                                    ,size=(5))
# half of students were assigned a .05 probability to eventually add up to whole sum of 1,
# the rest were .1 for the same reason. the ultimate goal however was to get a value
# for everyone to be randomly selected.

print(call_on_weighted)
```

## Generating a Normal Distribution

Create a variable named bell_curve and assign it a random distribution for
which the mean (peak of the curve) is 10, the standard deviation is 5, and the size of
the returned array is 4 rows by 5 columns. Print bell_curve.

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```

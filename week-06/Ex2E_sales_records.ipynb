This notebook analyzes a small sales dataset using Python functions.
Each function takes the full dataset as its first argument.

```python
sales_data = [
('Laptop', 'East', 1200.00),
('Phone', 'West', 850.00),
('Tablet', 'East', 400.00),
('Monitor', 'South', 650.00),
('Keyboard', 'West', 75.00),
('Laptop', 'South', 1150.00),
('Phone', 'East', 900.00),
('Tablet', 'West', 425.00),
('Monitor', 'East', 680.00),
('Keyboard', 'South', 80.00),
('Laptop', 'West', 1300.00),
('Phone', 'West', 820.00),
]
```

## Lab 1. Total sales by region

```python
def region_total(data, region):
    return sum(amt for _, r, amt in data if r == region)     # Returning total sales of region specified
              #get/add amt from tuples with 3 entries (_, r, amt) FROM data if r placement matches our region variable
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

## Lab 2. Count of sales by region

```python
def region_count(data, region):
    return sum(1 for _, r, _ in data if r == region)
#add 1 from tuples with 3 entries (_, r, amt) FROM data if r placement matches our region variable
```

```python
print(region_count(sales_data, 'East'))
print(region_count(sales_data, 'West'))
print(region_count(sales_data, 'South'))
```

## Lab 3. Average sale amount by region

```python
def region_average(data, region):
    if region_count(data, region) == 0:
        return 0
    else:
        return round(region_total(data, region) / region_count(data, region), 2)
```

```python
print(region_average(sales_data, 'East'))
print(region_average(sales_data, 'West'))
print(region_average(sales_data, 'South'))
print(region_average(sales_data, 'North'))
```

## Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):
    # Creating a new list that holds all data that has amounts larger than threshold
    new_list = []

    # Looping through index, checking amt column for values larger than threshold
    for i in range(len(data)):
        if data[i][2] > threshold: # if amt in record larger than threshold, add to new list
            new_list.append(data[i])
    
    return new_list # Returning new list containing records where sale amt is greater than threshold
```

```python
for entry in sales_above(sales_data, 500):
    print(entry)
```

```python
def sales_below(data, threshold):
    # Creating a new list that holds all data that has amounts smaller than threshold
    new_list = []

    # Looping through index, checking amt column for values smaller than threshold
    for i in range(len(data)):
        if data[i][2] < threshold: # if amt in record smaller than threshold, add to new list
            new_list.append(data[i])
    
    return new_list # Returning new list containing records where sale amt is smaller than threshold
```

```python
for entry in sales_below(sales_data, 500):
    print(entry)
```

## Lab 5. Top and bottom sale

```python
def top_sale(data): # returns the single record (tuple) with the highest sale amount
    return max(data, key=lambda s: s[2]) 
#Tuple is first placed in our key lamda function and it returns the 3rd slot (sales amount) and uses that to compare to other tuples for max value


def bottom_sale(data): # returns the single record (tuple) with the lowest sale amount
    return min(data, key=lambda s: s[2])
```

```python
t_prod, t_region, t_sales = top_sale(sales_data)
print(f'The top sale was a {t_prod} in the {t_region} region for ${t_sales}.')
```

## Lab 6. Apply a discount

```python
def apply_discount(data, discount_pct):
    return list(map(lambda s: (s[0], s[1], s[2] * (1 - discount_pct)), data))
```

```python
print(sales_data[:3])
print(apply_discount(sales_data,.15)[:3])
```

## Lab 7. Unique regions

```python
def unique_regions(data):
    return sorted(set(r for _, r, _ in data))
```

```python
print(unique_regions(sales_data))
```

In large datasets where thousands of values are repeatedly used (i.e., shipping data with countries, cities, etc.), this function is useful for identifying the distinct categories that make up our dataset. By extracting these unique values, we can easily group, summarize, or filter the records, essentially allowing us to see the structure hidden within overwhelming amounts of data.

## Lab 8. Region summary

```python
def region_summary(data, region):

    if region_count(data, region) == 0:
        return print(f'No data exists for the {region} region')

    top_record = top_sale([record for record in data if record[1] == region])

    print(f'=== {region} Region ===' '\n'
          f'Records: {region_count(data, region)}' '\n'
          f'Total Sales: ${region_total(data, region)}' '\n'
          f'Average Sale: ${region_average(data, region)}' '\n'
          f'Top Sale: {top_record[0]} -- ${top_record[2]}')
```

```python
region_summary(sales_data, 'East')
region_summary(sales_data, 'West')
region_summary(sales_data, 'South')
region_summary(sales_data, 'North')
```

## Lab 9. Full summary

```python
for region in unique_regions(sales_data):
        region_summary(sales_data, region)
```

# Today

6. In a code cell, define a variable called today that represents the current date and time

```python
today = datetime.datetime.now()
```

7. Add another code cell. Using the .strftime() method, identify the following pieces
of information and print the output:\
Today is [day of week], [month] [day], [year]

```python
print(f'Today is {today.strftime("%A")}, {today.strftime("%B")} {today.strftime("%d")}, {today.strftime("%Y")}')
```

# My Birthday

9. In the next cell, define a variable for your birthday using datetime.date().

```python
birthday = datetime.date(2001, 5, 21)
```

10. In the next cell, again using the .strftime() method, create the following output:

```python
print(f'My birthday is {birthday.strftime("%m")}/{birthday.strftime("%d")}/{birthday.strftime("%Y")}')
```

# Next 90 Days

12. In the next cell, define a variable for ninety_d which will be used to represent the
date 90 days from today

```python
ninety_d = today + datetime.timedelta(days=90)
```

13. In the next cell, again using the .strftime() method as needed, print the following
output: \
90 days from today is [month] [day], [year]

```python
print(f'90 days from today is {ninety_d.strftime("%B")} {ninety_d.strftime("%d")}, {ninety_d.strftime("%Y")}')
```

# Time for Dinner

15. Define a variable for dinner_time using datetime.time().

```python
dinner_time = datetime.time(17,30,0)
```

16. Again using the .strftime() method, create the following output:\
Let’s meet for dinner at HH:MM PM

```python
print(f"Let's meet for dinner at {dinner_time.strftime("%I")}:{dinner_time.strftime("%M %p")}")
```

2. Use separate code cells to create three functions: \
a) get_soc_sec_tax() accepts a gross pay amount and returns the Social Security
tax on that amount. Assume a tax rate of 6.2%

```python
def get_soc_sec_tax(gross_amount):
    #Multiplying gross amount with tax rate
    sst_amount = round(gross_amount * .062, 2)

    return sst_amount
```

b) get_medicare_tax() accepts a gross pay amount and returns the Medicare tax
on that amount. Assume a tax rate of 1.45%

```python
def get_medicare_tax(gross_amount):
    mt_amount = round(gross_amount * .0145, 2)
    return mt_amount
```

c) get_federal_tax() accepts a gross pay amount and withholding code, and
returns the federal tax withholding on that amount. Use the tax table below. (You
will need to use an if / else statement. The case for withholding code 4+ is extra tricky!)
Withholding Code \
 Tax Rate \
0   
 23% \
1   
 21% \
2   
 19.5% \
3   
 18.5% \
4+   18% less 0.5% for each withholding over 4 \
(ex, code 6: rate .18 – (2 x .005) = .17 (or 17%)

```python
def get_federal_tax(gross_amount, code):
    tax_rate = 0
    if code == 0:
        tax_rate = .23
    elif code == 1:
        tax_rate = .21
    elif code == 2:
        tax_rate = .195
    elif code == 3:
        tax_rate = .185
    elif code >= 4:
        tax_rate = .18 - ((code - 4) * .005)
    
    fed_tax_withheld = gross_amount * tax_rate
    return fed_tax_withheld
```

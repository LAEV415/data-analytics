# Exercise 2.A Using built-in functions
## Lab 1. Working with names – parsing strings

```python
test_name = 'James Bond'
```

```python
def parse_and_display(name):
    #loop through string name and find spaces
    for i in range(len(name)): # range returns all numbers within range value i.e 0 1 2 3 4 for range(5)
        if name[i] == ' ':
            first = name[:i] # :i --> 0 to before i
            last = name[i+1:] # i+1: --> everything after the space
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
            break
```

```python
parse_and_display(test_name)
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(test_list[1])
```

```python
for i in range(len(test_list)):
    parse_and_display(test_list[i])
```

```python
parse_and_display('Luis Espinosa')
```

4. In this lab, you will create multiple functions that, together, can parse a part code\
which is provided in the following format:\
supplier:productcode-size\
Examples:\
ACME:123-L\
 supplied by Acme, part #123, size large\
DI:12345-M\
 supplied by DI, part #12345, size medium\
ACE:1-12\
 supplied by Ace, part #1, size 12

5. In a separate code cell for each function, create the following: \
get_supplier_code()  returns all characters before the : \
get_product_num()  returns all characters between the : and -\
get_size()  returns all characters after the -

```python
def get_supplier_code(x):
    for i in range(len(x)):
        if x[i] == ':':
            supp_code = x[:i]
            return supp_code
```

```python
def get_product_num(x):
    start = 0
    end = 0
    for i in range(len(x)):
        if x[i] == ':':
            start = i + 1
        elif x[i] == '-':
            end = i
            break
    prod_num = x[start:end]
    return prod_num
```

```python
def get_size(x):
    for i in range(len(x)):
        if x[i] == '-':
            size = x[i+1:]
            return size
```

6. Once you have your functions ready, in another code cell create variables to hold the
three example part codes from above (if you haven’t already done so).

```python
a = 'ACME:123-L'
b = 'DI:12345-M'
c = 'ACE:1-12'
supp_code = None
prod_num = None
size = None
```

7. One at a time in new code cells for each test example, call all three functions for each
part code and use print statements to output the results as follows: \
For [full product code], the supplier is [supplier]. Part number [part #] is available in
size [size].

```python
supp_code = get_supplier_code(a)
prod_num = get_product_num(a)
size = get_size(a)
print(f'For {a}, the supplier is {supp_code}. Part number {prod_num} is available in size {size}')
```

```python
supp_code = get_supplier_code(b)
prod_num = get_product_num(b)
size = get_size(b)
print(f'For {b}, the supplier is {supp_code}. Part number {prod_num} is available in size {size}')
```

```python
supp_code = get_supplier_code(c)
prod_num = get_product_num(c)
size = get_size(c)
print(f'For {c}, the supplier is {supp_code}. Part number {prod_num} is available in size {size}')
```

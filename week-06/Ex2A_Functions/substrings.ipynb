6. In the next cell, write a function named parse_and_display that takes one
parameter, name. In the function, loop over the name and find the space, then extract the first and last name based on string indexing. Then add a print statement to see the results.

```python
def parse_and_display(name):
    #loop through string name and find spaces
    for i in range(len(name)): # range returns all numbers within range value i.e 0 1 2 3 4 for range(5)
        if name[i] == ' ':
            first = name[:i] # :i --> 0 to before i
            last = name[i+1:] # i+1: --> everything after the space
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
            break
```

11. CHALLENGE: In a new cell at the bottom, create a modified version of your function
called parse_and_display2() that will work with names made of one, two or
three words. Output should include middle name if applicable.
Some more example names to use when testing your new function: \
Lorde\
Billie Eilish\
Megan Thee Stallion

```python
def parse_and_display2(name):
    #Variable declaration
    first = None
    middle = None
    last = None
    space_index = 0
    #loop through string name and find spaces
    for i in range(len(name)): # range returns all numbers within range value i.e 0 1 2 3 4 for range(5)
        if name[i] == ' ':
            if space_index == 0:
                first = name[:i] # :i --> 0 to before i
                last = name[i+1:] # i+1: --> everything after the space
                space_index = i
            else:
                middle = name[space_index+1:i]
                last = name[i+1:]
                
    # In case full name is only one name            
    if space_index == 0:
        print(f"Full name: {name}")
    elif middle != None:
        print(f"Full name: {name}")
        print(f"First name: {first}")
        print(f'Middle name: {middle}')
        print(f"Last name: {last}")
    else:
        print(f"Full name: {name}")
        print(f"First name: {first}")
        print(f"Last name: {last}")
    
```

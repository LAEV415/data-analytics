4. Make your first update: Mark just reported a name change. \

```python
def update_list(employee_id, original, updated): #Use unique ID in case of repeated values (names, salary, etc)
    employee_id = str(employee_id)
    original = str(original)
    updated = str(updated)
    id_match = 0
    value_match = 0
    for i in range(len(hr_list)): #loop through list of tuples
        if hr_list[i][0] == employee_id: #If first entry matches id
            id_match = 1
            for j in range(len(hr_list[i])): #Loop through entries in tuple
                if hr_list[i][j] == original: #if info that needs to be updated exists
                    value_match = 1
                    hr_list[i] = list(hr_list[i]) #Change tuple list making it mutable
                    hr_list[i][j] = updated #Change matched entry into updated entry
                    hr_list[i] = tuple(hr_list[i]) #change back into tuple
                    break
            break
    if id_match == 0:
        print('ID does not exist in database.')
    elif id_match == 1 and value_match == 0:
        print('Original value does not exist and therefore cannot be updated.')

#Can improve by having user input what they would like to update and therefore not have them enter pre-existing information in case they don't know it

#Code to update multiple values at once
# def update_list(original_tuple, updated_tuple): #Enter existing tuple and updated tuple
#     for i in range(len(hr_list)): #loop through list of tuples
#         if hr_list[i] == original_tuple: #If inputted info matches a tuple
#             hr_list[i] = updated_tuple #Replace matched tuple with updated
#             break #Stops looping

```
